# Data cleaning

```python
# Make A Copy of dataset
df_ml = df.copy()
```

```python
# Cleaning "DATE OCC" column

from datetime import datetime

# custom parser for multiple formats (because there are multiple formates in column like - and / formates)
def parse_date(x):
    if pd.isna(x) or str(x).strip() == "":
        return pd.NaT
    for fmt in ("%m-%d-%Y %H:%M", "%m/%d/%Y %I:%M:%S %p", "%m/%d/%Y %H:%M"):
        try:
            return datetime.strptime(str(x), fmt)
        except:
            continue
    return pd.NaT

# Convert all to datetime
df_ml['DATE OCC'] = df_ml['DATE OCC'].apply(parse_date)
df_ml['Date Rptd'] = df_ml['Date Rptd'].apply(parse_date)

# Check how many NaT remain (for debugging)
print("Unparsed dates:", df_ml['DATE OCC'].isna().sum())
print("Unparsed dates:", df_ml['Date Rptd'].isna().sum())
```

```python
# Create new columns From DATE OCC foe Deep EDA

df_ml['YEAR OCC'] = df_ml['DATE OCC'].dt.year
df_ml['MONTH OCC'] = df_ml['DATE OCC'].dt.month
df_ml['DAY OCC'] = df_ml['DATE OCC'].dt.day
df_ml['WEEKDAY OCC'] = df_ml['DATE OCC'].dt.day_name()
```

```python
# Convert 'TIME OCC' column in to Numeric 

df_ml['TIME OCC'] = pd.to_numeric(df_ml['TIME OCC'], errors='coerce')
df_ml.loc[(df_ml['TIME OCC'] < 0) | (df_ml['TIME OCC'] > 2359), 'TIME OCC'] = None

# Now make a New Fature with just HOUR (For better analysis)

df_ml['HOUR OCC'] = (df_ml['TIME OCC'] // 100).astype('Int64')  
df_ml['HOUR OCC'] = df_ml['HOUR OCC'].replace(24, 0)
```

```python
df_ml.columns
```

```python
# Making a new Feature name "Delay Days" for EDA

df_ml['Delay Days'] = (df_ml['Date Rptd'] - df_ml['DATE OCC']).dt.days
```

```python
# Clean Vict Age Column

# Convert Nagative Values into Nan
df_ml.loc[df_ml['Vict Age'] < 0, 'Vict Age'] = None

# Very unrealistic ages convert into Nan( > 100 )
df_ml.loc[df_ml['Vict Age'] > 100, 'Vict Age'] = None

# Convert to integer
df_ml['Vict Age'] = df_ml['Vict Age'].astype('Int64')
```

```python
# --- Cleaning and Categorizing Victim Age ---

# Treat "0" values as Unknown (most likely missing age data)
df_ml['Vict Age'] = df_ml['Vict Age'].replace(0, pd.NA)

# Keep only realistic ages (1–100), set others to NaN
df_ml.loc[(df_ml['Vict Age'] < 1) | (df_ml['Vict Age'] > 100), 'Vict Age'] = pd.NA

# Create Age Group categories
bins = [0, 17, 30, 45, 60, 100]
labels = ["0-17 (Minor)", "18-30 (Young Adult)", "31-45 (Adult)", 
          "46-60 (Middle Age)", "61-100 (Senior)"]

df_ml['Vict Age Group'] = pd.cut(df_ml['Vict Age'], bins=bins, labels=labels, right=True)

# Fill missing values with "Unknown"
df_ml['Vict Age Group'] = df_ml['Vict Age Group'].cat.add_categories("Unknown").fillna("Unknown")

# Final check of Age Group distribution
print(df_ml['Vict Age Group'].value_counts())
```

```python
df_ml['Vict Sex'].value_counts(dropna=False)
```

```python
# --- Cleaning Vict Sex Column ---
# Replace inconsistent values with "Unknown"
df_ml['Vict Sex'] = df_ml['Vict Sex'].replace({
    'X': 'Unknown',   # "X" represents unspecified gender
    'H': 'Unknown',   # Rare miscoded values treated as Unknown
    '-': 'Unknown'    # Clearly erroneous entry
})

# Fill missing values with "Unknown"
df_ml['Vict Sex'] = df_ml['Vict Sex'].fillna('Unknown')

# Standardize valid categories for clarity
df_ml['Vict Sex'] = df_ml['Vict Sex'].replace({
    'M': 'Male',
    'F': 'Female'
})

# Final check of value distribution after cleaning
print(df_ml['Vict Sex'].value_counts())
```

```python
# Cleaning and Expanding Victim Descent
# Handling Null Values
# Mapping dictionary for Victim Descent codes

descent_map = {
    'A': 'Other Asian',
    'B': 'Black',
    'C': 'Chinese',
    'D': 'Cambodian',
    'F': 'Filipino',
    'G': 'Guamanian',
    'H': 'Hispanic/Latin/Mexican',
    'I': 'American Indian/Alaskan Native',
    'J': 'Japanese',
    'K': 'Korean',
    'L': 'Laotian',
    'O': 'Other',
    'P': 'Pacific Islander',
    'S': 'Samoan',
    'U': 'Hawaiian',
    'V': 'Vietnamese',
    'W': 'White',
    'Z': 'Asian Indian',
    'X': 'Unknown'   # Already Unknown in source
}

# Fill NaN with "X" (Unknown) before mapping
df_ml['Vict Descent'] = df_ml['Vict Descent'].fillna('X')

# Create descriptive column
df_ml['Vict Des Desc'] = df_ml['Vict Descent'].map(descent_map).fillna('Unknown')
```

# Data Preprocessing For Model Building

```python
# OUTLIRE HANDLING

# Before removing, check the min-max range of LAT/LON
print("LAT range:", df_ml['LAT'].min(), "to", df_ml['LAT'].max())
print("LON range:", df_ml['LON'].min(), "to", df_ml['LON'].max())

# Define valid range for Los Angeles
lat_min, lat_max = 33.0, 34.5
lon_min, lon_max = -119.0, -117.0

# Count how many rows are outside the range
invalid_rows = df_ml[
    (df_ml['LAT'] < lat_min) | (df_ml['LAT'] > lat_max) |
    (df_ml['LON'] < lon_min) | (df_ml['LON'] > lon_max)
]

print("Invalid LAT/LON rows:", len(invalid_rows))

# Remove those invalid rows
df_ml = df_ml[
    (df_ml['LAT'] >= lat_min) & (df_ml['LAT'] <= lat_max) &
    (df_ml['LON'] >= lon_min) & (df_ml['LON'] <= lon_max)
]

print("New Size after removing outliers:", df_ml.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Columns to scale
scale_cols = ['LAT', 'LON']

scaler = StandardScaler()
df_ml[scale_cols] = scaler.fit_transform(df_ml[scale_cols])

print("Scaling applied to:", scale_cols)
```

```python
# Sampling
```

```python
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans

# Stratified Sampling by AREA
# Taking a stratified sample of ~10% of the data.
# This ensures that the distribution of the 'AREA' column is preserved in the sample.

df_sample, _ = train_test_split(
    df_ml, 
    test_size=0.90,   # keep 10% data for analysis
    stratify=df_ml['AREA'], 
    random_state=42
)

print("Original shape:", df_ml.shape)
print("Sample shape:", df_sample.shape)

# Elbow Method on Sample

# Use only location features (latitude & longitude) for clustering.
# Compute WCSS (Within-Cluster Sum of Squares) for k = 1 to 10 clusters.
X_sample = df_sample[['LAT', 'LON']]  

wcss = []
K = range(1, 11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_sample)
    wcss.append(kmeans.inertia_)

# Plot Elbow Curve
# Plot the Elbow curve to visually determine the optimal number of clusters.
plt.figure(figsize=(8,5))
plt.plot(K, wcss, 'bo-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('WCSS (Within-Cluster Sum of Squares)')
plt.title('Elbow Method (Stratified Sample)')
plt.show()
```

# K-MEANS

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import pandas as pd

# Cluster with KMeans on ORIGINAL lat/lon
# ==========================
k = df_ml[['LAT', 'LON']]  # 

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_ml['K_Cluster'] = kmeans.fit_predict(k)

# Density per cluster
cluster_density = df_ml.groupby('K_Cluster').size().sort_values(ascending=False)
print("Cluster densities:\n", cluster_density)
```

Evalution Metric For Cheking CLuter Quality.

silhoutte = 0.5 - 1 = good clustet (high=best).

Davies-Bouldin = near to 0 to 0.5 is good  (low=Best).

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score

# Model Trained on full dataset (df['Cluster'])
k_full = df_ml[['LAT','LON']]

# Sample for silhouette
k_sample = k_full.sample(n=20000, random_state=42)  # 20k rows for Faster Computation
labels_sample = df_ml.loc[k_sample.index, 'K_Cluster']

sil_score = silhouette_score(k_sample, labels_sample)
print("Silhouette Score (sampled 20k):", sil_score)

# Davies-Bouldin (full data)
db_score = davies_bouldin_score(k_full, df_ml['K_Cluster'])
print("Davies-Bouldin Score (full data):", db_score)
```

```python
# PLot 
top_points = []
for cluster in df_ml['K_Cluster'].unique():
    cluster_data = df_ml[df_ml['K_Cluster'] == cluster]

    grouped = cluster_data.groupby([cluster_data['LAT'].round(3),
                                    cluster_data['LON'].round(3)]).size().reset_index(name='count')

    top10 = grouped.nlargest(10, 'count').copy()
    top10['K_Cluster'] = cluster
    top_points.append(top10)

top_points = pd.concat(top_points)

# Severity assign (1 = highest risk, 5 = lowest risk)
top_points['Severity'] = top_points.groupby('K_Cluster')['count'].rank(ascending=False, method='first').astype(int)

# Severity color mapping
severity_colors = {
    1: 'red',
    2: 'orange',
    3: 'yellow',
    4: 'cyan',
    5: 'blue',
    6: 'pink',
    7: 'white',
    8: 'gray',
    9: 'brown',
    10: 'magenta'
}

# Plot clusters + Risky hotspots
plt.figure(figsize=(12, 10))

# Cluster scatter (background)
plt.scatter(df_ml['LON'], df_ml['LAT'],
            c=df_ml['K_Cluster'], cmap='viridis', s=8, alpha=0.4)

# Plot top hotspots
for severity, color in severity_colors.items():
    subset = top_points[top_points['Severity'] == severity]
    plt.scatter(subset['LON'], subset['LAT'],
                c=color, s=80, edgecolor='black', linewidth=0.8,
                label=f"{severity}")

# ==========================
# 🎯 Custom legend format
# ==========================
handles, labels = plt.gca().get_legend_handles_labels()

# Legend banate waqt ek "heading" add karte hain
from matplotlib.legend import Legend
plt.legend(handles, labels, title="Risk Levels", title_fontsize=12, fontsize=10)

# Final plot settings
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Crime Hotspot Detection with Risk Levels")
plt.show()
```

# DBSCAN

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np
import matplotlib.pyplot as plt

# same Lat and Lon
X = df_sample[['LAT', 'LON']] # HERE WE TOOK SAMPLE DATA BECAUSE DBSCAN MODEL TAKE HEAVY TIME SO...

# getting Nearest Neighbors for selcetion of eps
neigh = NearestNeighbors(n_neighbors=5)
nbrs = neigh.fit(X)
distances, indices = nbrs.kneighbors(X)

# distances ko sort karo
distances = np.sort(distances[:, 4])  # 4th neighbor (n_neighbors-1)
plt.plot(distances)
plt.xlabel("Points")
plt.ylabel("5th Nearest Distance")
plt.title("K-distance Graph for DBSCAN")
plt.show()
```

```python
# fit model on data and making cluster

from sklearn.cluster import DBSCAN

db = df_sample[['LAT', 'LON']]

# DBSCAN model
dbscan = DBSCAN(eps=0.12, min_samples=200)  
clusters = dbscan.fit_predict(db)

# Cluster labels ko dataset me add karo
df_sample['Db_Cluster'] = clusters

# Output check karo
print(df_sample['Db_Cluster'].value_counts())
```

```python
# Removing Noice
mask = df_sample['Db_Cluster'] != -1
filtered_data = df_sample[mask][['LAT','LON']]
filtered_labels = df_sample[mask]['Db_Cluster']

# Check IF 2 OR MORE CLUSTE WAS MADE
if len(set(filtered_labels)) > 1:
   
    # Silhouette Score (sampled)
    sample_size = min(20000, len(filtered_data))
    sil_score = silhouette_score(
        filtered_data, 
        filtered_labels, 
        sample_size=sample_size, 
        random_state=42
    )
    # Davies-Bouldin (full data)
    db_score = davies_bouldin_score(filtered_data, filtered_labels)

    print(f"Silhouette Score (sampled {sample_size}): {sil_score:.2f}")
    print(f"Davies-Bouldin Index: {db_score:.2f}")
```

```python
# Hotspot severity ranking
top_points = []
for cluster in df_sample['Db_Cluster'].unique():
    if cluster == -1:  # noise skip karna
        continue
    cluster_data = df_sample[df_sample['Db_Cluster'] == cluster]

    grouped = cluster_data.groupby([cluster_data['LAT'].round(3),
                                    cluster_data['LON'].round(3)]).size().reset_index(name='count')

    top10 = grouped.nlargest(10, 'count').copy()
    top10['Cluster'] = cluster
    top_points.append(top10)

top_points = pd.concat(top_points)

# Severity assign (1 = highest risk, 10 = lowest risk per cluster)
top_points['Risk'] = top_points.groupby('Cluster')['count'].rank(ascending=False, method='first').astype(int)

# Severity color mapping
severity_colors = {
    1: 'red',
    2: 'orange',
    3: 'yellow',
    4: 'cyan',
    5: 'blue',
    6: 'pink',
    7: 'white',
    8: 'gray',
    9: 'brown',
    10: 'magenta'
}

# Plot clusters + noise + severity hotspots
plt.figure(figsize=(12, 10))

# Cluster colormap (tab10 for clear distinct colors)
clusters = [c for c in set(df_sample['Db_Cluster']) if c != -1]
cmap = plt.cm.get_cmap("tab10", len(clusters))

# Plot clusters (without noise)
for idx, cluster in enumerate(clusters):
    subset = df_sample[df_sample['Db_Cluster'] == cluster]
    plt.scatter(subset['LON'], subset['LAT'],
                c=[cmap(idx)], s=8, alpha=0.3)

# Plot noise separately in gray
noise = df_sample[df_sample['Db_Cluster'] == -1]
if not noise.empty:
    plt.scatter(noise['LON'], noise['LAT'], c='black', s=6, alpha=0.2)

# Plot top hotspots (severity points)
for severity, color in severity_colors.items():
    subset = top_points[top_points['Risk'] == severity]
    plt.scatter(subset['LON'], subset['LAT'],
                c=color, s=80, edgecolor='black', linewidth=0.8,
                label=f"{severity}")

#Legend (Risk levels)
handles, labels = plt.gca().get_legend_handles_labels()

# Remove cluster and Noice
filtered_handles_labels = [(h, l) for h, l in zip(handles, labels) if l.isdigit()]

# Legend draw
plt.legend(*zip(*filtered_handles_labels),
           title="Risk Levels", title_fontsize=12, fontsize=10)

# Final plot settings

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Crime Hotspot Detection (DBSCAN) with Risk Levels")
plt.show()
```

# HDBSCAN

```python
import hdbscan

# Data select
hd = df_sample[['LAT', 'LON']]

# HDBSCAN clustering
# ==========================
hdbscan_model = hdbscan.HDBSCAN(
    min_cluster_size=400,
    min_samples=400,
    cluster_selection_epsilon=0.2,
    metric='euclidean'
)

df_sample['HDBSCAN_Cluster'] = hdbscan_model.fit_predict(hd)

# Cluster summary
print("Cluster Distribution:")
print(df_sample['HDBSCAN_Cluster'].value_counts())

# Evaluation Metrics
labels = df_sample['HDBSCAN_Cluster']

# removing Noise (-1)
mask = labels != -1
db_clustered = hd[mask]
labels_clustered = labels[mask]

# Check: scores only print if cluster are made more than 1
if len(set(labels_clustered)) > 1:
    # Silhouette Score on sample (20k points max)
    sample_size = min(20000, len(db_clustered))
    idx = np.random.choice(len(db_clustered), size=sample_size, replace=False)
    sil_score = silhouette_score(db_clustered.iloc[idx], labels_clustered.iloc[idx])

    # Davies-Bouldin Index on full clustered data
    db_score = davies_bouldin_score(db_clustered, labels_clustered)

    print(f"\nSilhouette Score (sampled {sample_size}): {sil_score:.3f}")
    print(f"Davies-Bouldin Index (full data): {db_score:.3f}")
```

```python
# Hotspot severity ranking

top_points = []

for cluster in df_sample['HDBSCAN_Cluster'].unique():
    if cluster == -1:  # noise skip
        continue

    cluster_data = df_sample[df_sample['HDBSCAN_Cluster'] == cluster]

    grouped = cluster_data.groupby([cluster_data['LAT'].round(3),
                                    cluster_data['LON'].round(3)]).size().reset_index(name='count')

    top10 = grouped.nlargest(10, 'count').copy()
    top10['Cluster'] = cluster
    top_points.append(top10)

top_points = pd.concat(top_points)

# Severity assign (1 = highest risk, 10 = lowest risk per cluster)
top_points['Risk'] = top_points.groupby('Cluster')['count'].rank(
    ascending=False, method='first'
).astype(int)

# Risk color mapping
risk_colors = {
    1: 'red',
    2: 'orange',
    3: 'yellow',
    4: 'cyan',
    5: 'blue',
    6: 'pink',
    7: 'white',
    8: 'gray',
    9: 'brown',
    10: 'magenta'
}

# Plot clusters + noise + severity hotspots
plt.figure(figsize=(12, 10))

clusters = [c for c in set(df_sample['HDBSCAN_Cluster']) if c != -1]
cmap = plt.cm.get_cmap("tab10", len(clusters))

# Plot clusters (without noise)
for idx, cluster in enumerate(clusters):
    subset = df_sample[df_sample['HDBSCAN_Cluster'] == cluster]
    plt.scatter(subset['LON'], subset['LAT'],
                c=[cmap(idx)], s=8, alpha=0.4)

# Plot noise separately in gray
noise = df_sample[df_sample['HDBSCAN_Cluster'] == -1]
if not noise.empty:
    plt.scatter(noise['LON'], noise['LAT'], c='black', s=6, alpha=0.3, label="Noise")

# Plot top hotspots (risk points)
for risk, color in risk_colors.items():
    subset = top_points[top_points['Risk'] == risk]
    plt.scatter(subset['LON'], subset['LAT'],
                c=color, s=80, edgecolor='black', linewidth=0.8,
                label=f"{risk}")

# Legend
handles, labels = plt.gca().get_legend_handles_labels()
risk_handles = [h for h, l in zip(handles, labels) if l.isdigit()]
risk_labels = [l for l in labels if l.isdigit()]

plt.legend(risk_handles, risk_labels,
           title="Risk Levels", title_fontsize=12, fontsize=10)

# Final plot settings
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Crime Hotspot Detection (HDBSCAN) with Risk Levels")
plt.show()
```

# Tuning

```python
X = df_sample[['LAT', 'LON']]

# Parameter grid
param_grid = {
    "min_cluster_size": [200, 300, 400, 500, 800],
    "min_samples": [50, 100, 200, 400],
    "cluster_selection_epsilon": [0.05, 0.1, 0.2, 0.3]
}

results = []

for mcs in param_grid["min_cluster_size"]:
    for ms in param_grid["min_samples"]:
        for eps in param_grid["cluster_selection_epsilon"]:
            model = hdbscan.HDBSCAN(
                min_cluster_size=mcs,
                min_samples=ms,
                cluster_selection_epsilon=eps,
                metric='euclidean'
            )
            
            labels = model.fit_predict(X)

            # Noise exclude
            mask = labels != -1
            X_clustered = X[mask]
            labels_clustered = pd.Series(labels[mask])

            # only Print metrics if cluster made more than 1 
            if len(set(labels_clustered)) > 1:
                sample_size = min(20000, len(X_clustered))
                idx = np.random.choice(len(X_clustered), size=sample_size, replace=False)

                sil_score = silhouette_score(X_clustered.iloc[idx], labels_clustered.iloc[idx])
                db_score = davies_bouldin_score(X_clustered, labels_clustered)

                results.append({
                    "min_cluster_size": mcs,
                    "min_samples": ms,
                    "epsilon": eps,
                    "clusters": len(set(labels_clustered)),
                    "noise_points": (labels == -1).sum(),
                    "silhouette": sil_score,
                    "davies_bouldin": db_score
                })

# Results DataFrame
df_results = pd.DataFrame(results).sort_values(by=["silhouette"], ascending=False)
print(df_results.head(10))  # top 10 configs
```

After tuning the HDBSCAN model, the best parameters were found to be:

Minimum Cluster Size = 800

Minimum Samples = 400

Epsilon = (auto-determined by the algorithm)

With these settings, the results are as follows:

Silhouette Score: 0.554134

Davies-Bouldin Index: 0.561317

Compared to the baseline model, the Davies-Bouldin score remained the same, 
while the Silhouette Score improved only marginally (by ~0.001). 
This slight change suggests that the model was already close to its optimal configuration, 
and further tuning of hyperparameters yields limited additional benefits.
